```python
from PyPDF2 import PdfReader
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.feature_extraction import text
from sklearn.metrics.pairwise import cosine_similarity
from google.colab import drive
```

```python
# Path disesuaikan dengan direktori masing-masing
file_paths = [
    "/content/drive/MyDrive/Pattern Recognition/Paper/paper1.pdf",
    "/content/drive/MyDrive/Pattern Recognition/Paper/paper2.pdf",
    "/content/drive/MyDrive/Pattern Recognition/Paper/paper3.pdf"
]

# Ambil teks dari sejumlah halaman dokumen
documents = []
for path in file_paths:
    reader = PdfReader(path)
    full_text = ""
    for page in reader.pages[:1]: # Ubah [:1] sesuai dengan jumlah halaman yang akan diproses
        page_text = page.extract_text()
        if page_text:
          full_text += page_text + "\n"  # Teks halaman ke-n digabung menjadi 1 ke variabel full_text
    documents.append(full_text.strip())
```

```python
stopwords_indonesia = [
    'yang', 'dan', 'di', 'ke', 'dari', 'untuk', 'dengan', 'pada', 'adalah', 'itu',
    'ini', 'sebagai', 'bahwa', 'dalam', 'oleh', 'atau', 'juga', 'sudah', 'karena',
    'lebih', 'akan', 'bisa', 'tidak', 'maka', 'namun', 'agar', 'setelah', 'telah',
    'menjadi', 'tersebut', 'bagi', 'selama', 'antara', 'sehingga', 'hanya', 'kita',
    'mereka', 'kami', 'saya', 'ia', 'dia', 'apa', 'siapa', 'dimana', 'kapan',
    'bagaimana', 'adapun', 'masih', 'perlu', 'tanpa',
]

stopwords_inggris = text.ENGLISH_STOP_WORDS

# Menggabungkan stopwords bahasa Indonesia dan bahasa Inggris
stopwords = stopwords_indonesia+list(stopwords_inggris)

# Hitung TF-IDF
vectorizer = TfidfVectorizer(stop_words=stopwords)
tfidf_matrix = vectorizer.fit_transform(documents)

# Hitung cosine similarity
similarity_matrix = cosine_similarity(tfidf_matrix)

# Tampilkan hasil
print("Cosine Similarity Matrix (paper 1-3):")
for i in range(len(similarity_matrix)):
    for j in range(len(similarity_matrix)):
        print(f"Similarity(paper{i+1}, paper{j+1}) = {similarity_matrix[i][j]:.4f}")
```
